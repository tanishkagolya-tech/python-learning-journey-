```python
class Bank:

    def __init__(self, balance):

        self.balance = balance

    def withdraw(self, amount):

        if amount < 0:
            raise Exception("amount can't be negative")

        if amount > self.balance:
            raise Exception("you do not have sufficient balance")

        self.balance = self.balance - amount


obj = Bank(1000)

try:

    obj.withdraw(150000)

except Exception as e:

    print(e)

else:

    print(obj.balance)
```

```python
class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print('logout')

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError('Alert! device change')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('login error')



obj = Google('tanishka','tg@gmail.com','1234','android')

try:
  obj.login('tg@gmail.com','1234','windows')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print('database connection closed')

```
