````python
# can super access parent ka data?
# using super outside the class
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        print(super().brand)

s=SmartPhone(20000, "Apple", 13)

s.buy()
````

In inheritance, super() is mainly used to access parent class methods or constructors, not instance variables directly. In my code, brand is an instance variable created using self.brand = brand inside the constructor. Instance variables belong to the object (s) and are stored inside the current object’s memory, not inside the parent class itself. That is why super().brand gives an error — because super() looks for class-level methods in the parent class, while brand is object data. Since the SmartPhone object already inherited the brand attribute from Phone, the correct way to access it is through self.brand. Think of it like this: super() is used to call parent functionality (like methods), while self is used to access the current object’s data and attributes

````python
# super -> constuctor
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
        print ("Inside smartphone constructor")

s=SmartPhone(20000, "Samsung", 12, "Android", 2)

print(s.os)
print(s.brand)
````

This code demonstrates how inheritance and super() work with constructors in Python. The Phone class is the parent class, and the SmartPhone class is the child class that inherits from Phone. The main purpose of this code is to show how a child class can reuse the constructor of the parent class instead of rewriting the same code again.

This code demonstrates how inheritance and `super()` work with constructors in Python. The `Phone` class is the parent class, and the `SmartPhone` class is the child class that inherits from `Phone`. The main purpose of this code is to show how a child class can reuse the constructor of the parent class instead of rewriting the same code again.

First, the parent class is created:


class Phone:


This class acts like a blueprint for all phone objects.

Inside it, the constructor is defined:


def __init__(self, price, brand, camera):


The constructor runs automatically whenever a `Phone` object (or child object) is created. It accepts three values: `price`, `brand`, and `camera`.

The next line:

print("Inside phone constructor")


simply prints a message so we can see when the parent constructor is executed.

Then these lines create instance variables:


self.__price = price
self.brand = brand
self.camera = camera

here - 
* `self.__price` is a private instance variable because of double underscore `__`
* `self.brand` and `self.camera` are public instance variables
* These variables are stored inside the object being created

After that, the child class is created:


class SmartPhone(Phone):

This means:

SmartPhone IS-A Phone
```

So the child class inherits all properties and methods of the `Phone` class.

Now the child class constructor is defined:

def __init__(self, price, brand, camera, os, ram):


This constructor accepts all parent data plus two extra values:

* `os`
* `ram`

The line:


print('Inside smartphone constructor')
```

runs first when a `SmartPhone` object is created.

Then comes the MOST IMPORTANT line:


super().__init__(price, brand, camera)


This calls the parent (`Phone`) constructor. Without this line, the child class would not get `brand`, `camera`, or `price` attributes. The `super()` keyword is used to access parent class functionality. Here it tells Python to execute the constructor of `Phone`.

When this line runs, Python internally does something similar to:

Phone.__init__(self, price, brand, camera)
```

So now the child object receives all parent variables:

* `brand`
* `camera`
* `__price`

After returning from the parent constructor, execution comes back to the child constructor and these lines run:


self.os = os
self.ram = ram

These are new instance variables specific to `SmartPhone`.

Then this line executes:


print("Inside smartphone constructor")


Finally, the object is created:


s = SmartPhone(20000, "Samsung", 12, "Android", 2)


Python follows this flow:

1. Create `SmartPhone` object
2. Run child constructor
3. Child constructor calls parent constructor using `super()`
4. Parent constructor creates phone-related attributes
5. Control returns to child constructor
6. Child-specific attributes (`os`, `ram`) are created

The object `s` now contains:


brand = Samsung
camera = 12
os = Android
ram = 2


Then:


print(s.os)


prints:

Android


because `os` belongs directly to the child class.

And:


print(s.brand)

prints:

Samsung


because `brand` was inherited from the parent class through `super()`.

The biggest concept in this code is that inheritance allows code reuse. Instead of rewriting the `price`, `brand`, and `camera` logic again in `SmartPhone`, the child class simply reuses the parent constructor using `super()`. This is one of the core ideas behind scalable and professional software design.

````python
# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
````

In this code, both `Phone` and `Product` classes have a method named `buy()`. Then `SmartPhone` inherits from both classes using multiple inheritance:


class SmartPhone(Phone, Product):


Now when we create an object and call:


s.buy()


Python gets confused about which `buy()` method should run — the one from `Phone` or the one from `Product`.

To solve this confusion, Python follows something called MRO (Method Resolution Order). MRO simply means the order in which Python searches classes for methods.

Python always searches from left to right.

So in:


class SmartPhone(Phone, Product):

the search order becomes:


SmartPhone → Phone → Product


When `s.buy()` runs:

1. Python first checks inside `SmartPhone`
2. No `buy()` method is found there
3. Then it checks `Phone`
4. `Phone` already has a `buy()` method
5. So Python executes that method and stops searching

That’s why output becomes:

Buying a phone


The `Product.buy()` method never runs because Python already found the method in `Phone`.

This searching order is called MRO.


The output is:


Inside phone constructor
Buying a phone


because when the object of `SmartPhone` is created:

s = SmartPhone(20000, "Apple", 12)

Python first looks for a constructor (`__init__`) inside `SmartPhone`. Since `SmartPhone` does not have its own constructor, Python automatically uses the constructor of the first parent class, which is `Phone`.

So this line runs:

print("Inside phone constructor")
```

That’s why the first output is:

Inside phone constructor
```

After that, when:


s.buy()


runs, Python again searches for the `buy()` method.

Search order is:

SmartPhone → Phone → Product
```

`SmartPhone` does not have a `buy()` method, so Python checks `Phone` and finds:

def buy(self):
    print("Buying a phone")
So it executes that method and stops searching further.

That’s why second output becomes:


Buying a phone
