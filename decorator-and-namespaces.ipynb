### Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

A namespace is simply a container that stores names (variables, functions, classes) and their corresponding values.

Think of it as a dictionary behind the scenes.
There are 4 types of namespaces:
- Builtin Namespace
- Global Namespace
- Enclosing Namespace
- Local Namespace

```python
#Why do we need namespaces
x = 10

def func():
    x = 20

#Now there are two variables named x.

Which one should Python use?

Namespaces solve this confusion.
```

Namespace Search Rule (LEGB)

When Python sees a variable:

print(x)

it searches in this order:

L → Local

Current function

E → Enclosing

Outer function

G → Global

Current file

B → Built-in

Python's built-in functions

This is called the LEGB Rule.

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

# Decorators
A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python

Built in decorators like @staticmethod, @classmethod, @abstractmethod and @property etc
User defined decorators that we programmers can create according to our needs

A decorator takes a function,
adds extra behavior,
and returns a new function.

```python
# Python are 1st class function

def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,4)
```

```python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello riya')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```

```python

import time

def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print('time taken by',func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello wolrd')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

@timer
def power(a,b):
  print(a**b)

hello()
square(2)
power(2,3)
```

This code demonstrates the use of a decorator in Python. The timer() function acts as a decorator that measures the execution time of any function passed to it. Inside the decorator, the wrapper() function records the start time using time.time(), executes the original function, and then calculates the total time taken. The @timer syntax is used to apply the decorator to functions such as hello(), square(), and power(). Whenever these functions are called, the decorator automatically prints the function name and the time required for execution. This helps in monitoring and comparing the performance of different functions without modifying their original code.

Key Points
@timer applies the decorator to a function.

func(*args) executes the original function.

time.time() records the current time.

func.__name__ retrieves the function name.

*args allows the decorator to work with functions having different numbers of arguments.

Decorators help add extra functionality (like timing, logging, authentication, etc.) without changing the original function code.

```python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('This will not work')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)
```
