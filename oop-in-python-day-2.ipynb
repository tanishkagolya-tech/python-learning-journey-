Write OOP classes to handle the following scenarios:

A user can create and view 2D coordinates

A user can find out the distance between 2 coordinates

A user can find find the distance of a coordinate from origin

A user can check if a point lies on a given line

A user can find the distance between a given 2D point and a given line

```python
class Point:

    def __init__(self, x, y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return '<{},{}>'.format(self.x_cod, self.y_cod)

    def euclidean_distance(self, other):
        return ((self.x_cod - other.x_cod)**2 +
                (self.y_cod - other.y_cod)**2)**0.5

    def distance_from_origin(self):
        return (self.x_cod**2 + self.y_cod**2)**0.5


class Line:

    def __init__(self, A, B, C):
        self.A = A
        self.B = B
        self.C = C

    def __str__(self):
        return '{}x + {}y + {} = 0'.format(self.A, self.B, self.C)

    def point_on_line(line, point):

        if line.A * point.x_cod + line.B * point.y_cod + line.C == 0:
            return "lies on the line"

        else:
            return "does not lie on the line"
    def shortest_distance(line,point):
        return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5
    def intersection(self, other):

     if self.A / other.A != self.B / other.B:
        return "Lines intersect"

     elif (self.A / other.A == self.B / other.B ==
           self.C / other.C):
        return "Same line"

     else:
        return "Parallel lines"




p1 = Point(0,0)
p2 = Point(-2,3)

print(p1)
print(p2)
#<x,y>
p1.euclidean_distance(p2)
p1.distance_from_origin()


l1 = Line(2,7,9)
print(l1)

l1 = Line(1,1,-2)
p1 = Point(1,2)
print(l1)
print(p1)
l1.point_on_line(p1)

l1.shortest_distance(p1)

```

```python
class Rectangle:

    def __init__(self, length, width):

        self.length = length
        self.width = width

    def area(self):

        return self.length * self.width

    def perimeter(self):

        return 2 * (self.length + self.width)

    def square(self):

        if self.length == self.width:
            return "square"

        else:
            return "not square"
```

```python
r1 = Rectangle(10, 5)
```

```python
print(r1.area())

print(r1.perimeter())

print(r1.square())
```

```python
class Point:

    def __init__(self, x, y):

        self.x = x
        self.y = y

    def distance_from_origin(self):

        return (self.x**2 + self.y**2)**0.5


class Circle:

    def __init__(self, radius):

        self.radius = radius

    def area(self):

        return 3.14 * self.radius * self.radius

    def circumference(self):

        return 2 * 3.14 * self.radius

    def point_inside(self, point):

        if point.distance_from_origin() <= self.radius:
            return "inside"

        else:
            return "outside"


p1 = Point(2,3)

c1 = Circle(5)

print(c1.point_inside(p1))
```

```python
 class Atm:

    # constructor
    def __init__(self):

        self.pin = ''
        self.__balance = 0

        self.menu()

    # menu function
    def menu(self):

        user_input = input("""
Hi how can I help you?

1. Press 1 to create pin
2. Press 2 to change pin
3. Press 3 to check balance
4. Press 4 to withdraw
5. Press anything else to exit

""")

        if user_input == '1':

            self.create_pin()

        elif user_input == '2':

            self.change_pin()

        elif user_input == '3':

            self.check_balance()

        elif user_input == '4':

            self.withdraw()

        else:

            print("Thank you for using ATM")
            exit()

    # create pin
    def create_pin(self):

        user_pin = input('Enter your pin: ')
        self.pin = user_pin

        user_balance = int(input('Enter balance: '))
        self.__balance = user_balance

        print('PIN created successfully')

        self.menu()

    # change pin
    def change_pin(self):

        old_pin = input('Enter old pin: ')

        if old_pin == self.pin:

            new_pin = input('Enter new pin: ')
            self.pin = new_pin

            print('PIN changed successfully')

        else:

            print('Wrong PIN')

        self.menu()

    # check balance
    def check_balance(self):

        user_pin = input('Enter your pin: ')

        if user_pin == self.pin:

            print('Your balance is:', self.__balance)

        else:

            print('Invalid PIN')

        self.menu()

    # withdraw money
    def withdraw(self):

        user_pin = input('Enter your pin: ')

        if user_pin == self.pin:

            amount = int(input('Enter amount: '))

            if amount <= self.__balance:

                self.__balance = self.__balance - amount

                print('Withdrawal successful')
                print('Remaining balance:', self.__balance)

            else:

                print('Insufficient balance')

        else:

            print('Invalid PIN')

        self.menu()


# object creation
obj = Atm()
```

Static Variables(Vs Instance variables)

```python
class Atm:

    __counter = 1

    def __init__(self):

        self.pin = ''
        self.__balance = 0

        self.cid = Atm.__counter

        Atm.__counter += 1

    @staticmethod
    def get_counter():

        return Atm.__counter

    def get_balance(self):

        return self.__balance

    def set_balance(self, new_value):

        if type(new_value) == int:

            self.__balance = new_value

        else:

            print("Balance must be integer")

    def create_pin(self):

        self.pin = input("Enter pin: ")

        self.__balance = int(input("Enter balance: "))

        print("PIN created successfully")

    def change_pin(self):

        old_pin = input("Enter old pin: ")

        if old_pin == self.pin:

            self.pin = input("Enter new pin: ")

            print("PIN changed successfully")

        else:

            print("Wrong pin")

    def check_balance(self):

        user_pin = input("Enter pin: ")

        if user_pin == self.pin:

            print("Balance:", self.__balance)

        else:

            print("Invalid pin")

    def withdraw(self):

        user_pin = input("Enter pin: ")

        if user_pin == self.pin:

            amount = int(input("Enter amount: "))

            if amount <= self.__balance:

                self.__balance -= amount

                print("Withdrawal successful")

                print("Remaining balance:", self.__balance)

            else:

                print("Insufficient balance")

        else:

            print("Invalid pin")
```

```python
#static method
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
```

Static attributes are created at class level.

Static attributes are accessed using ClassName.

Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.

The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.
